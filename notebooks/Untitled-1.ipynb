```python
%pip install pandas numpy matplotlib scipy statsmodels ipykernel
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Find the Goldman project folder
root = Path.cwd()
if not (root / "data").exists():
    root = root.parent

# Load monthly stock prices
us_prices = pd.read_csv(root / "data" / "prices_us_monthly.csv")
sg_prices = pd.read_csv(root / "data" / "prices_sg_monthly.csv")

print("US data:", us_prices.shape)
print("Singapore data:", sg_prices.shape)

display(us_prices.head())
display(sg_prices.head())


# Identify the date column
print("Columns:", us_prices.columns.tolist())

# Display the full dataset preview
pd.set_option("display.max_columns", None)
display(us_prices.head())

print(us_prices.columns.tolist())
print(sg_prices.columns.tolist())
```

```python

# Prepare monthly returns
def prepare_returns(prices):
    df = prices.copy()
    df["Date"] = pd.to_datetime(df["Date"])
    df = df.set_index("Date").sort_index()
    df = df.apply(pd.to_numeric, errors="coerce")

    return df.pct_change(fill_method=None).replace(
        [np.inf, -np.inf], np.nan
    )

us_returns = prepare_returns(us_prices)
sg_returns = prepare_returns(sg_prices)

# Build two alternative long-short factors
def build_factors(returns):
    # Use 12 prior monthly returns to estimate volatility
    volatility = returns.rolling(
        window=12, min_periods=12
    ).std().shift(1)

    # Previous month's return, known before trading
    previous_return = returns.shift(1)

    def long_short(signal, long_low=True):
        factor = {}

        for date in returns.index:
            scores = signal.loc[date].dropna()
            valid = scores.index.intersection(
                returns.loc[date].dropna().index
            )

            if len(valid) < 6:
                factor[date] = np.nan
                continue

            ranked = scores.loc[valid].sort_values()
            n = max(1, len(ranked) // 3)

            low = ranked.index[:n]
            high = ranked.index[-n:]

            if long_low:
                long_assets, short_assets = low, high
            else:
                long_assets, short_assets = high, low

            factor[date] = (
                returns.loc[date, long_assets].mean()
                - returns.loc[date, short_assets].mean()
            )

        return pd.Series(factor)

    return pd.DataFrame({
        "LOWVOL": long_short(volatility),
        "REV": long_short(previous_return)
    })

us_factors = build_factors(us_returns)
sg_factors = build_factors(sg_returns)

print("US alternative factors:")
display(us_factors.dropna().head())

print("Singapore alternative factors:")
display(sg_factors.dropna().head())
```

```python
ff3 = pd.read_csv(root / "data" / "ff3_monthly.csv")

print("Fama-French columns:")
print(ff3.columns.tolist())

display(ff3.head())
```

```python

import statsmodels.api as sm

# Align Fama-French dates with monthly stock returns
ff = ff3.copy()
ff["Date"] = pd.to_datetime(ff["Date"]).dt.to_period("M")
ff = ff.set_index("Date").sort_index()
ff = ff[["Mkt-RF", "RF"]].apply(
    pd.to_numeric, errors="coerce"
)

# Align stock returns and constructed factors
returns = us_returns.copy()
returns.index = returns.index.to_period("M")

factors = us_factors.copy()
factors.index = factors.index.to_period("M")

# Combine all observations
data = returns.join(
    ff, how="inner"
).join(
    factors, how="inner"
)

results = []

for ticker in us_returns.columns:
    cols = [ticker, "RF", "Mkt-RF", "LOWVOL", "REV"]
    sample = data[cols].dropna()

    if len(sample) < 36:
        continue

    y = sample[ticker] - sample["RF"]

    capm = sm.OLS(
        y, sm.add_constant(sample[["Mkt-RF"]])
    ).fit()

    multi = sm.OLS(
        y,
        sm.add_constant(
            sample[["Mkt-RF", "LOWVOL", "REV"]]
        )
    ).fit()

    results.append({
        "Asset": ticker,
        "CAPM Alpha (%)": capm.params["const"] * 100,
        "Multifactor Alpha (%)": multi.params["const"] * 100,
        "CAPM |t(alpha)|": abs(capm.tvalues["const"]),
        "Multifactor |t(alpha)|": abs(multi.tvalues["const"]),
        "CAPM R2": capm.rsquared,
        "Multifactor R2": multi.rsquared
    })

alpha_results = pd.DataFrame(results).set_index("Asset")

display(alpha_results.round(3))

print("Assets analyzed:", len(alpha_results))
print("CAPM significant alphas:",
      (alpha_results["CAPM |t(alpha)|"] > 2).sum())
print("Multifactor significant alphas:",
      (alpha_results["Multifactor |t(alpha)|"] > 2).sum())

# Plot alpha comparison
ax = alpha_results[
    ["CAPM Alpha (%)", "Multifactor Alpha (%)"]
].plot(
    kind="bar",
    figsize=(16, 6),
    width=0.8
)

ax.axhline(0, color="black", linewidth=1)
ax.set_title("US: CAPM vs LOWVOL + REV Alphas")
ax.set_ylabel("Monthly Alpha (%)")
ax.set_xlabel("Asset")
plt.xticks(rotation=65, ha="right")
plt.tight_layout()
plt.show()
```

```python

# Load Singapore value weights
vw_sg = pd.read_csv(
    root / "data" / "value_weights_sg.csv"
)

print("Columns:")
print(vw_sg.columns.tolist())

print("\nFirst five rows:")
display(vw_sg.head())

print("\nShape:", vw_sg.shape)
```

```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import statsmodels.api as sm

# 1. Standardize monthly dates
returns = sg_returns.copy()
returns.index = pd.to_datetime(returns.index).to_period("M")

factors = sg_factors.copy()
factors.index = pd.to_datetime(factors.index).to_period("M")

rf = rf_sg.copy()
rf["month"] = pd.to_datetime(rf["Date"]).dt.to_period("M")
rf = rf.set_index("month")["RF"].astype(float)

# 2. Prepare time-varying market-cap weights
weights = vw_sg.copy()
weights["month"] = pd.to_datetime(
    weights["date"]
).dt.to_period("M")

weights = weights.pivot_table(
    index="month",
    columns="asset",
    values="market_cap",
    aggfunc="last"
)

# Shift market capitalizations one month backward
# so each return uses information from the prior month
weights = weights.shift(1)

# 3. Construct value-weighted market proxy
common_assets = returns.columns.intersection(weights.columns)

market_returns = []

for month in returns.index:
    if month not in weights.index:
        market_returns.append(np.nan)
        continue

    r = returns.loc[month, common_assets]
    caps = weights.loc[month, common_assets]

    valid = r.notna() & caps.notna() & (caps > 0)

    if valid.sum() < 6:
        market_returns.append(np.nan)
        continue

    w = caps[valid] / caps[valid].sum()
    market_returns.append((w * r[valid]).sum())

market = pd.Series(
    market_returns,
    index=returns.index,
    name="Market"
)

# 4. Combine returns, risk-free rate, and factors
data = returns.join(market).join(
    rf.rename("RF")
).join(factors)

data["Mkt-RF"] = data["Market"] - data["RF"]

# 5. Estimate both models on identical samples
results_sg = []

for ticker in common_assets:
    sample = data[
        [ticker, "RF", "Mkt-RF", "LOWVOL", "REV"]
    ].dropna()

    if len(sample) < 36:
        continue

    y = sample[ticker] - sample["RF"]

    capm = sm.OLS(
        y,
        sm.add_constant(sample[["Mkt-RF"]])
    ).fit()

    multi = sm.OLS(
        y,
        sm.add_constant(
            sample[["Mkt-RF", "LOWVOL", "REV"]]
        )
    ).fit()

    results_sg.append({
        "Asset": ticker,
        "CAPM Alpha (%)": capm.params["const"] * 100,
        "Multifactor Alpha (%)": multi.params["const"] * 100,
        "CAPM |t(alpha)|": abs(capm.tvalues["const"]),
        "Multifactor |t(alpha)|": abs(multi.tvalues["const"]),
        "CAPM R2": capm.rsquared,
        "Multifactor R2": multi.rsquared
    })

alpha_results_sg = pd.DataFrame(
    results_sg
).set_index("Asset")

display(alpha_results_sg.round(3))

print("Assets analyzed:", len(alpha_results_sg))
print(
    "CAPM significant alphas:",
    (alpha_results_sg["CAPM |t(alpha)|"] > 2).sum()
)
print(
    "Multifactor significant alphas:",
    (alpha_results_sg["Multifactor |t(alpha)|"] > 2).sum()
)

# 6. Plot alpha comparison
ax = alpha_results_sg[
    ["CAPM Alpha (%)", "Multifactor Alpha (%)"]
].plot(
    kind="bar",
    figsize=(16, 6),
    width=0.8
)

ax.axhline(0, color="black", linewidth=1)
ax.set_title(
    "Singapore: CAPM vs LOWVOL + REV Alphas"
)
ax.set_ylabel("Monthly Alpha (%)")
ax.set_xlabel("Asset")
plt.xticks(rotation=65, ha="right")
plt.tight_layout()
plt.show()
```
